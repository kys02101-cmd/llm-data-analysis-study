### 왜 이 질문을 선택했는가?
매출 데이터를 분석할 때 단순히 총매출만 확인하는 것보다 어떤 상품이 매출을 발생시키는지, 매출이 언제 증가하는지, 어떤 고객이 구매에 기여하는지를 함께 파악하는 것이 중요하다고 생각했습니다.

이에 상품 카테고리별 매출을 통해 주요 매출 발생 영역을 파악하고, 월별 매출을 통해 매출 변화의 원인을 살펴보고자 했습니다. 또한 고객별 구매 횟수와 누적 구매 금액을 비교하여 반복 구매 고객의 특징을 확인하고자 했습니다.

### 결과 관찰
카테고리별 매출에서는 패션이 32,940,000원으로 가장 높은 매출을 기록. 식품이 25,035,000원으로 그 다음이고, 전자기기와 도서도 비교적 높은 매출.

월별 매출은 2026년 8월에 16,980,000원으로 가장 높음. 2025년 9월에는 5,299,000원으로 가장 낮음. 월별 주문 건수와 평균 주문 금액을 함께 비교할 필요가 있음을 확인.

완료 주문 고객 99명 중 53명은 2회 이상 구매. 가장 높은 누적 구매 금액을 기록한 고객은 고객 ID 3으로, 3회 주문하여 총 6,702,000원을 구매.

### 나의 해석과 판단
상품 카테고리별 매출 차이는 판매 수량과 상품 단가의 영향을 함께 받는 것으로 보입니다. 패션과 식품은 판매 수량과 매출액 모두 높은 편이므로 주요 매출 카테고리로 볼 수 있습니다.

월별 매출은 주문 건수만으로 설명하기 어렵습니다. 주문 건수가 적더라도 평균 주문 금액이 높으면 전체 매출이 높아질 수 있으므로 두 지표를 함께 살펴보아야 합니다.

또한 반복 구매 고객이 전체 완료 주문 고객 중 절반 이상을 차지한다는 점에서 기존 고객의 재구매를 유도하는 전략도 중요하다고 판단했습니다.

### 업무·분석적 의미
카테고리별 매출 분석은 재고 및 마케팅 자원을 어느 상품군에 집중할지 결정하는 데 활용할 수 있습니다.

월별 매출 분석은 매출이 증가하거나 감소하는 시점을 파악하고, 프로모션이나 판매 전략의 효과를 추가로 검토하는 데 도움이 됩니다.

고객별 구매 분석은 반복 구매 고객을 구분하고, 고객의 구매 횟수와 구매 금액에 따라 차별화된 고객 관리 전략을 수립하는 기초 자료로 활용할 수 있습니다.

### 한계와 추가 확인 사항
상품 데이터와 연결되지 않는 주문상품 16건의 매출 4,015,000원이 존재합니다. 따라서 카테고리별 매출 합계는 전체 완료 주문 매출과 일치하지 않습니다.

주문월이 누락된 데이터의 매출은 1,903,000원입니다. 월별 매출 추이를 해석할 때 해당 금액이 제외되지 않도록 주의해야 합니다.

고객 데이터와 연결되지 않는 고객 ID가 1개 존재하므로 고객별 분석 결과에도 일부 누락 가능성이 있습니다.

## 3. 관찰 → 가설 → 추가 검증
### 사례 1
- 관찰: 패션 카테고리가 32,940,000원으로 가장 높은 매출을 기록했다.
- 가설: 패션 상품은 다른 카테고리보다 판매 수량이 많거나 상품 단가가 높아 매출 기여도가 높을 수 있다.
- 추가 검증: 카테고리별 평균 판매 단가, 상품별 판매 수량, 구매 고객 수를 비교하여 매출 차이가 판매량과 단가 중 어떤 요인에서 비롯되었는지 확인한다.

### 사례 2
- 관찰: 2026년 8월 매출은 16,980,000원으로 분석 기간 중 가장 높았으며, 2025년 9월 매출은 5,299,000원으로 가장 낮았다.
- 가설: 월별 매출 차이는 주문 건수뿐 아니라 고액 주문의 발생 여부에도 영향을 받았을 수 있다.
- 추가 검증: 월별 주문 건수와 평균 주문 금액을 비교하고, 월별 상위 주문의 구매 상품 및 금액을 확인하여 매출 증가 원인을 분석한다.

### 사례 3
- 관찰: 완료 주문 고객 99명 중 53명이 2회 이상 구매했으며, 가장 높은 누적 구매 금액을 기록한 고객은 3회 주문하여 6,702,000원을 구매했다.
- 가설: 반복 구매 고객은 일회성 구매 고객보다 누적 구매 금액이 높을 가능성이 있다.
- 추가 검증: 반복 구매 고객과 일회성 구매 고객의 평균 누적 구매 금액 및 평균 주문 금액을 비교하고, 재구매 간격과 구매 카테고리도 함께 분석한다.

## 4. LLM 질문 확장
- Safe Context: 고객 개인정보 및 이름은 제외하고, 고객 ID, 상품 카테고리, 주문일, 주문 상태, 판매 수량, 매출액 등 분석에 필요한 집계 데이터만 활용한다.
- Prompt:
"온라인 쇼핑몰의 고객, 상품, 주문, 주문상품 데이터를 활용하여 EDA를 진행하고 있다. 완료 주문을 기준으로 카테고리별 매출, 월별 매출, 고객별 구매 횟수를 분석했다. 패션 카테고리의 매출이 가장 높았으며, 2026년 8월 매출이 가장 높게 나타났다. 완료 주문 고객 중 반복 구매 고객도 상당수 존재한다. 현재 분석에서 놓쳤을 수 있는 변수나 추가로 설정할 만한 EDA 질문을 제안해 줘. 각 질문은 관찰 가능한 지표와 검증 방법을 함께 제시하고, 데이터만으로 확인할 수 없는 인과관계는 단정하지 말아 줘."
- LLM 제안 요약:
고객 연령대별 선호 카테고리와 구매 금액을 비교한다.
결제 수단별 평균 주문 금액과 구매 빈도를 비교한다.
반복 구매 고객과 일회성 구매 고객의 구매 특성을 비교한다.
월별 매출 변화와 카테고리별 매출 변화의 관계를 확인한다.
- 실제 사용 여부: 수정 후 사용
- 판단 이유: LLM이 제안한 질문은 기존 분석을 확장하는 데 유용하지만, 실제 데이터에 해당 변수가 존재하는지와 집계 결과가 충분한지 먼저 확인해야 한다. 또한 단순한 상관관계를 인과관계로 해석하지 않도록 주의할 필요가 있다.
- 사람이 수정한 내용: 고객 이름 등 개인정보는 분석 대상에서 제외하고 고객 ID를 활용하도록 했다. 또한 완료 주문을 기준으로 분석하도록 범위를 명확히 했으며, 실제 데이터에서 확인할 수 있는 지표와 추가 검증 방법을 중심으로 질문을 수정했다.

## 5. 다음 분석 우선순위
1. 고객별 구매 횟수와 누적 구매 금액을 비교하여 반복 구매 고객의 특성을 분석한다.

2. 월별 주문 건수와 평균 주문 금액을 비교하여 매출 변화의 원인을 분석한다.

3. 카테고리별 평균 판매 단가와 판매 수량을 비교하여 매출 차이가 발생한 원인을 분석한다.

### 가장 먼저 확인하고 싶은 이유
고객별 구매 특성을 가장 먼저 확인하고 싶습니다. 완료 주문 고객 중 2회 이상 구매한 고객이 53명으로 나타났기 때문에 반복 구매 고객이 전체 매출에 얼마나 기여하는지 분석할 필요가 있습니다. 이를 통해 재구매 유도 및 고객 유지 전략에 활용할 수 있는 근거를 마련하고자 합니다.

### 현재 데이터만으로 단정할 수 없는 것
현재 데이터만으로는 특정 카테고리의 매출이 높은 원인이나 특정 월의 매출이 증가한 이유를 확정할 수 없습니다. 또한 반복 구매 고객의 구매 금액이 높은 이유가 충성도 때문인지, 상품 단가나 구매 시점의 영향 때문인지도 판단하기 어렵습니다.
따라서 추가적인 고객 행동 데이터, 프로모션 정보, 상품별 가격 및 구매 이력 등을 함께 분석해야 보다 정확한 결론을 도출할 수 있다고 보았습니다.
