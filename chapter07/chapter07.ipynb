### 원본 집계값
데이터 범위는 completed 주문 상세 474행을 상품별로 합산한 집계임. 완료 주문이 한 번이라도 있는 상품 98개가 점으로 표시됨.

| 항목 | 값 |
| --- | ---: |

| 가격과 판매 수량의 상관계수(피어슨 / 스피어만) | -0.24 / -0.27 |

가격 구간별 요약

판매 수량이 가장 많은 상품은 가격 163,000원에 35개(판매 금액 5,705,000원).

### 결과 관찰
- 점들은 가격 5,000원~200,000원 전 구간에 넓게 흩어져 있고, 직선이나 곡선 모양은 보이지 않습니다.
- 판매 수량은 대부분 2~35개 사이이며 중앙값은 14개입니다.
- 가격이 높은 구간으로 갈수록 판매 수량이 조금씩 낮아집니다. 가격 구간별 평균 판매 수량은 16.4개, 16.7개, 15.2개, 12.0개이고, 상관계수는 -0.24(피어슨), -0.27(스피어만)로 약한 음의 값입니다.
- 판매 수량이 30개 이상인 상품은 4개이며, 가격은 71,000원~163,000원으로 서로 다릅니다.

### 나의 해석과 판단
가격과 판매 수량 사이에는 약한 음의 관계가 보이지만 "가격이 높으면 덜 팔린다"고 할 만큼 뚜렷하지는 않습니다. 상관계수 -0.24를 제곱하면 약 0.06이어서 가격으로 설명되는 수량 변동은 6% 정도에 그칩니다. 같은 가격대에서도 수량이 2개인 상품과 30개인 상품이 함께 있으므로, 수량의 차이는 가격보다 다른 요인이 더 크게 좌우할 가능성이 큽니다. 다만 가격 구간별 평균을 보면 15만 원 이상 구간의 판매 수량(12.0개)이 5만 원 미만 구간(16.4개)보다 낮아서, 약한 경향 자체는 있다고 판단했습니다. 이 구간별 차이는 제가 임의로 나눈 구간에 따라 달라질 수 있어서 참고 정도로만 보았습니다.

### 업무·분석적 의미
수량만 보면 저가 상품이 조금 더 많이 팔리는 것처럼 보이지만, 금액으로 보면 반대입니다. 15만 원 이상 구간의 상품당 평균 판매 금액은 2,088,219원으로 5만 원 미만 구간(457,222원)의 약 5배입니다. 따라서 판매 수량을 늘리려는 목적과 판매 금액을 늘리려는 목적에서 보는 상품이 달라질 수 있고, 상품 구성이나 프로모션을 정할 때 두 지표를 함께 봐야 합니다.

### 한계와 추가 확인 사항
- 상관이 약하게 보이는 것일 뿐 가격이 수량의 원인이라고 할 수 없습니다. 할인, 노출 위치, 재고, 상품이 판매된 기간 같은 변수는 이 데이터에 없습니다.
- 점 하나는 상품 하나이고, 상품당 완료 주문 수의 중앙값이 5건 정도라 점마다 표본이 작아 우연의 영향을 받을 수 있습니다.
- 카테고리별로 패턴이 다를 수 있는데 이 그래프는 카테고리를 구분하지 않았습니다.
- 완료 주문만 사용했으므로 취소·환불된 주문의 수요는 반영되지 않았습니다.

### 업무·분석적 의미
상위 고객을 한 덩어리로 보기보다 반복 구매형(4~5건, 평균 주문 금액 낮음)과 고액 단건형(2건, 평균 주문 금액 높음)으로 나누어 접근할 수 있습니다. 예를 들어 반복 구매형은 재구매 관리, 고액 단건형은 다음 구매를 이끄는 방법을 따로 검토할 수 있습니다.

### 한계와 추가 확인 사항
- 분석 기간이 약 1년이고 고객마다 가입 시점이 달라서, 구매 금액 차이에 가입 후 경과 기간이 섞여 있습니다.
- 완료 주문만 사용했으므로 취소·환불이 많은 고객은 따로 확인하지 못했습니다. 구매 금액이 이익 기여를 뜻하지도 않습니다.

### 결과 관찰
- completed가 184건으로 가장 많고, cancelled 64건, refunded 52건 순입니다.
- completed는 cancelled의 약 2.9배, refunded의 약 3.5배입니다.
- cancelled와 refunded를 합치면 116건으로 전체 주문의 38.67%입니다. 10건 중 약 4건이 완료되지 않았습니다.
- cancelled가 refunded보다 12건 많습니다.

### 나의 해석과 판단
전체 주문의 약 39%가 완료되지 않았다는 점이 이번 시각화에서 가장 중요한 사실이라고 판단했습니다. 이 비율이 높기 때문에 금액을 볼 때는 반드시 completed로 범위를 정해야 합니다. 상태를 구분하지 않고 합계를 내면 255,610,000원이 되어 completed 금액보다 약 41.7% 과대하게 보입니다. 그래서 앞의 카테고리·월별·고객 금액 그래프는 모두 completed 범위를 사용해야 한다는 것을 이 그래프가 뒷받침한다고 보았습니다.

### 업무·분석적 의미
매출성 지표를 만들 때 어떤 상태를 포함할지 기준이 없으면 숫자가 40% 가까이 달라질 수 있습니다. 또한 취소와 환불이 각각 21%, 17%라는 것은 운영 측면에서 줄일 여지가 있는 규모라서, 원인을 확인하는 분석을 따로 해 볼 가치가 있습니다.

### 한계와 추가 확인 사항
- 상태별 건수만으로는 취소와 환불의 원인을 알 수 없습니다.
- 이 그래프에는 시점이 없어서 최근으로 갈수록 취소가 늘었는지 줄었는지 알 수 없습니다.
- 결제수단 같은 다른 변수는 이 그래프에서 확인할 수 없습니다. 별도로 계산해 보면 취소 비율이 카드 25.7%, 네이버페이 19.5%로 차이가 났지만, 결제수단별 주문 수가 70~79건이라 표본이 작아 우연일 수 있어서 근거로 쓰지는 않았습니다.

## 6. Chapter 07 최종 판단
### 가장 중요한 그래프 1개와 이유
주문 상태별 주문 수 막대그래프(대표 그래프 3)입니다. 이 그래프가 다른 모든 금액 분석의 범위를 정해 주기 때문입니다. 전체 주문의 약 39%(116건)가 취소·환불이었고, 상태를 구분하지 않으면 금액이 약 41.7% 과대하게 계산됩니다. 그래프 자체는 단순하지만 이 사실을 모르면 이후의 카테고리·월별·고객 그래프를 잘못된 범위로 읽을 수 있습니다.

### 그래프가 보여 준 사실
- 전체 주문 300건 중 completed 184건(61.33%), cancelled 64건(21.33%), refunded 52건(17.33%)입니다.
- 상품 가격과 완료 주문 판매 수량 사이에는 약한 음의 관계(상관계수 -0.24)가 있으며, 가격이 같은 상품들 사이의 수량 차이가 가격에 따른 차이보다 큽니다.
- 완료 구매 금액 상위 10명은 4,100,000원에서 2,990,000원 사이로 격차가 작고, 완료 주문 전체 금액의 23.29%를 차지합니다.
- 상위 10명에는 반복 구매형(4~5건)과 고액 단건형(2건)이 섞여 있습니다.

### 그래프만으로 말할 수 없는 것
- 가격이 판매 수량을 결정하는지, 즉 원인 관계는 알 수 없습니다.
- 상위 고객이 충성 고객인지, 앞으로도 구매할 고객인지는 알 수 없습니다.
- 취소와 환불이 왜 일어났는지, 시간이 지나면서 늘었는지 줄었는지는 알 수 없습니다.
